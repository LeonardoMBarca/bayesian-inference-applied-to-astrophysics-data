# Modelo Bayesiano M1 - HAT-P-7 b

Notebook reprodutível do primeiro modelo bayesiano preliminar do projeto.

Este notebook executa o mesmo pipeline do script `scripts/run_bayesian_baseline.py`. O modelo M1 é um baseline simples do tipo box transit e não deve ser interpretado como caracterização física final do planeta.

## Execução

A lógica principal fica em `src/bayesian_modeling/legacy/baseline.py`. O notebook carrega o entry point compatível, executa `run_pipeline()` e mostra os principais artefatos gerados.

```python
import importlib.util
import sys
from pathlib import Path

cwd = Path.cwd().resolve()
PROJECT_ROOT = cwd if (cwd / "data" / "gold").exists() else cwd.parent
script_path = PROJECT_ROOT / "scripts" / "run_bayesian_baseline.py"

spec = importlib.util.spec_from_file_location("bayesian_baseline", script_path)
bayesian_baseline = importlib.util.module_from_spec(spec)
sys.modules[spec.name] = bayesian_baseline
spec.loader.exec_module(bayesian_baseline)

results = bayesian_baseline.run_pipeline(project_root=PROJECT_ROOT)
results
```

## Tabelas Posteriores

As tabelas abaixo são salvas em `tables/bayesian_baseline/hat_p_7_b/`.

```python
import pandas as pd

posterior_summary = pd.read_csv(PROJECT_ROOT / results["posterior_summary_path"])
derived_summary = pd.read_csv(PROJECT_ROOT / results["derived_summary_path"])
posterior_summary, derived_summary
```
