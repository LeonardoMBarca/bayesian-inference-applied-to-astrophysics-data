## Execução reprodutível

A lógica principal está em `src/gold_analysis/lightcurve.py`. O notebook carrega o entry point compatível `scripts/analyze_gold_lightcurve.py` e executa a mesma função usada pela linha de comando.

```python
import importlib.util
import sys
from pathlib import Path

cwd = Path.cwd().resolve()
PROJECT_ROOT = cwd if (cwd / "data" / "gold").exists() else cwd.parent
script_path = PROJECT_ROOT / "scripts" / "analyze_gold_lightcurve.py"

spec = importlib.util.spec_from_file_location("gold_eda", script_path)
gold_eda = importlib.util.module_from_spec(spec)
sys.modules[spec.name] = gold_eda
spec.loader.exec_module(gold_eda)

results = gold_eda.run_analysis(project_root=PROJECT_ROOT)
results
```
